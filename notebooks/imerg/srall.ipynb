```python
import os
import json
import warnings
import numpy as np
import pandas as pd
import xarray as xr
import proplot as pplt
from matplotlib.lines import Line2D
from biascorrect import load_pairs,fit_qm
warnings.filterwarnings('ignore')
pplt.rc.update({
    'savefig.dpi':900,
    'savefig.bbox':'tight',
    'savefig.pad_inches':0.02,
    'tick.minor':False,
    'font.size':9,
    'label.size':9,
    'tick.labelsize':9,
    'title.size':9,
    'abc.size':9,
    'legend.fontsize':9,
    'suptitle.size':9,
    'leftlabelsize':9,
    'toplabelsize':9,
    'leftlabel.weight':'normal',
    'toplabel.weight':'normal',
    'reso':'xx-hi'})
```

```python
with open('../../scripts/configs.json','r',encoding='utf-8') as f:
    CONFIGS = json.load(f)
SPLITSDIR  = CONFIGS['filepaths']['splits']
WEIGHTSDIR = CONFIGS['filepaths']['weights']
MODELSDIR  = CONFIGS['filepaths']['models']
PREDSDIR   = CONFIGS['filepaths']['predictions']
FIELDVARS  = CONFIGS['experiments']['nn']['runs']['nn_gauss']['fieldvars']
SEEDS      = CONFIGS['experiments']['nn']['seeds']
EQUATIONS  = CONFIGS['experiments']['sr']['optimizedeqs']
LATRANGE   = CONFIGS['domain']['latrange']
LONRANGE   = CONFIGS['domain']['lonrange']
SPLIT      = 'test'
CONSTNAMES = {'sr_all_eq':['c9','c10'],'sr_all_pc_eq':['c12','c13']}
PCNAME     = 'sr_all_pc_eq'
NBINS      = 20
MINSAMPLES = 50
MINPRECIP  = 0.1
LANDFRAC   = 0.5
NHOURBINS  = 8
with open(os.path.join(SPLITSDIR,'stats.json'),'r',encoding='utf-8') as f:
    STATS = json.load(f)
REGISTRY   = {row['name']:json.loads(row['constants']) for _,row in pd.read_csv(os.path.join(MODELSDIR,'sr','optimized_equations.csv')).iterrows()}
NQUANTILES = 200
FITSPLITS  = ['train','valid']
FIGSDIR    = 'figs'
os.makedirs(FIGSDIR,exist_ok=True)
```

```python
def to_precip(exponent):
    return np.maximum(np.expm1(exponent),0.0)

def calc_correction(thetaeanom,shf,lf,lamshf,lfc,pc):
    return lamthetae*((lfc-lf)**3+pc*(1-lfc)**3)*thetaeanom+lamshf*(lfc-lf)**3*(shf-STATS['shf_mean'])

def calc_diurnal_cycle(z,mask):
    counts = np.bincount(hourbin[mask],minlength=NHOURBINS)
    cycle  = np.bincount(hourbin[mask],weights=z[mask],minlength=NHOURBINS)/np.maximum(counts,1)
    return np.r_[cycle[-1],cycle,cycle[0]]

def calc_binned_mean(z):
    return np.where(counts>=MINSAMPLES,np.bincount(binidx,weights=z[oceanmask],minlength=NBINS**2)/np.maximum(counts,1),np.nan).reshape(NBINS,NBINS).T
```

```python
mapping = fit_qm(*load_pairs(SPLITSDIR,FITSPLITS),nquantiles=NQUANTILES)
c3,c4,c5 = (REGISTRY['sr_atm_eq'][name] for name in ['c3','c4','c5'])
lam      = STATS['tp_std']*c3/STATS['thetae_std']**3
gamma    = c4*STATS['thetae_std']/STATS['thetaestar_std']
thetac   = STATS['thetae_mean']-gamma*STATS['thetaestar_mean']+c5*STATS['thetae_std']
kappa    = STATS['thetae_std']/STATS['rh_std']
lamthetae = STATS['tp_std']/STATS['thetae_std']
params    = {name:(STATS['tp_std']*REGISTRY[name][cshf]/STATS['shf_std'],REGISTRY[name][clfc],float(name==PCNAME)) for name,(cshf,clfc) in CONSTNAMES.items()}

with xr.open_dataset(os.path.join(SPLITSDIR,f'{SPLIT}.h5'),engine='h5netcdf') as ds:
    ntime,nlat,nlon = (ds.sizes[dim] for dim in ['time','lat','lon'])
    lat,lon,dsig    = ds['lat'].values,ds['lon'].values,ds['dsig'].values
    hours  = ds['time'].dt.hour.values
    fields = np.stack([ds[name].transpose('time','lat','lon','sig').values.reshape(-1,dsig.size) for name in FIELDVARS],axis=1)
    tp,pr,shf = (ds[name].transpose('time','lat','lon').values.ravel() for name in ['tp','pr','shf'])
    lf     = xr.broadcast(ds['lf'],ds['tp'])[0].transpose('time','lat','lon').values.ravel().astype(np.float64)
savedtp = {}
for name in CONSTNAMES:
    with xr.open_dataset(os.path.join(PREDSDIR,f'{name}_{SPLIT}_predictions.nc')) as ds:
        savedtp[name] = ds['tp'].squeeze().transpose('time','lat','lon').values.ravel()
kernels = []
for seed in SEEDS:
    with xr.open_dataset(os.path.join(WEIGHTSDIR,f'nn_gauss_{seed}_weights.nc'),engine='h5netcdf') as ds:
        kernels.append(ds['k'].values)

integrals   = dict(zip(FIELDVARS,(fields*np.mean(kernels,axis=0)[None,:,:]*dsig[None,None,:]).sum(axis=2).T))
moisture    = kappa*(integrals['rh']-STATS['rh_mean'])
instability = integrals['thetae']-gamma*integrals['thetaestar']-thetac
thetaeanom  = integrals['thetae']-STATS['thetae_mean']
atmexp      = lam*np.maximum(moisture,instability)**3
atmtp       = to_precip(atmexp)
corrections = {name:calc_correction(thetaeanom,shf,lf,*param) for name,param in params.items()}
predtp      = {name:to_precip(atmexp+correction) for name,correction in corrections.items()}
pc2grad     = {name:np.where(moisture>=instability,0.0,3*lam*instability**2)+lamthetae*((lfc-lf)**3+pc*(1-lfc)**3) for name,(_,lfc,pc) in params.items()}
finite      = np.isfinite(moisture)&np.isfinite(instability)&np.isfinite(corrections[PCNAME])&np.isfinite(tp)
pctp        = predtp[PCNAME]
for name in CONSTNAMES:
    print(f"Max difference from saved {EQUATIONS[name]['description']} predictions: {np.nanmax(np.abs(predtp[name][finite]-savedtp[name][finite])):.2e} mm")
    print(f"{EQUATIONS[name]['description']} PC2 satisfaction (all grid points): {100*np.mean(pc2grad[name][finite]>=0):.1f}%")
atmpr       = mapping(atmtp)
pcpr        = mapping(pctp)
finite      = finite&np.isfinite(pr)

shape       = (ntime,nlat,nlon)
localhour   = (hours[:,None,None]+lon[None,None,:]/15.0)%24
hourbin     = (np.broadcast_to(localhour,shape).ravel()//(24/NHOURBINS)).astype(int)
hourcenters = (np.arange(NHOURBINS)+0.5)*24/NHOURBINS
cyclehours  = np.r_[hourcenters[-1]-24,hourcenters,hourcenters[0]+24]
regionmasks = {'Ocean':lf<LANDFRAC,'Land':lf>=LANDFRAC}
gain        = np.nanmean(np.where(finite,(atmpr-pr)**2-(pcpr-pr)**2,np.nan).reshape(shape),axis=0)
cycles      = {(region,source):calc_diurnal_cycle(values,finite&regionmask)
               for region,regionmask in regionmasks.items() for source,values in [('IMERG',pr),('SR-ATM',atmpr),('SR-ALL-PC',pcpr)]}

oceanmask = finite&(atmpr>=MINPRECIP)&regionmasks['Ocean']
edges     = (np.linspace(*np.percentile(shf[oceanmask],[1,99]),NBINS+1),np.linspace(*np.percentile(thetaeanom[oceanmask],[1,99]),NBINS+1))
centers   = [0.5*(edge[:-1]+edge[1:]) for edge in edges]
binidx    = np.clip(np.digitize(shf[oceanmask],edges[0])-1,0,NBINS-1)*NBINS+np.clip(np.digitize(thetaeanom[oceanmask],edges[1])-1,0,NBINS-1)
counts    = np.bincount(binidx,minlength=NBINS**2)
made      = calc_binned_mean(pcpr-atmpr)
needed    = calc_binned_mean(pr-atmpr)
```

```python
colors  = {'IMERG':'gray','SR-ATM':EQUATIONS['sr_atm_eq']['color'],'SR-ALL-PC':EQUATIONS['sr_all_pc_eq']['color']}
styles  = {'Ocean':'-','Land':':'}
handles = [Line2D([],[],color=color,linewidth=2,label=label) for label,color in colors.items()]
handles+= [Line2D([],[],color='k',linestyle=linestyle,linewidth=2,label=region) for region,linestyle in styles.items()]

fig,axs = pplt.subplots([[1,2],[3,4]],refwidth=2.4,hratios=(1,1.5),proj={1:'cyl'},share=False,span=False,wspace=1.2,hspace=6)
axs[0].format(title='SR-ALL-PC Improvement Over SR-ATM',coast=True,lonlim=LONRANGE,lonlines=[65,75,85],lonlabels='b',latlim=LATRANGE,latlines=[10,15,20],latlabels='l')
axs[1].set_box_aspect((LATRANGE[1]-LATRANGE[0])/(LONRANGE[1]-LONRANGE[0]))
axs[1].format(title='Diurnal Cycle of Precipitation',xlim=(0,23),xticks=[6,12,18],xticklabels=['06 LST','12 LST','18 LST'],
              ylabel='Precipitation Rate (mm/hr)',ylim=(0,2),yticks=0.5,ytickloc='right',yticklabelloc='right',ylabelloc='right')
for ax in axs[2:]:
    ax.set_box_aspect(1)
    ax.format(xlabel='SHF (W/m$^2$)',xlim=(edges[0][0],edges[0][-1]),ylim=(edges[1][0],edges[1][-1]))
axs[2].format(title='SR-ALL-PC $-$ SR-ATM (Ocean)',ylabel='$\\widehat{\\mathit{\\theta_e}}-\\mathit{\\mu_{\\theta_e}}$ (K)')
axs[3].format(title='IMERG $-$ SR-ATM (Ocean)',yticklabels=[])

mgain = axs[0].pcolormesh(lon,lat,gain,cmap='ColdHot_r',vmin=-2,vmax=2,levels=16,extend='both')
axs[0].colorbar(mgain,loc='b',ticks=1,label='$\\Delta$MSE (mm$^2$/hr$^2$)')
for (region,source),cycle in cycles.items():
    axs[1].plot(cyclehours,cycle,color=colors[source],linestyle=styles[region],linewidth=2)
axs[1].legend(handles,loc='b',ncols=3,columnspacing=0.6,handlelength=1.5,handletextpad=0.4)
mcorr = axs[2].pcolormesh(*centers,made,cmap='ColdHot_r',vmin=-4,vmax=4,levels=32,extend='both')
axs[3].pcolormesh(*centers,needed,cmap='ColdHot_r',vmin=-4,vmax=4,levels=32,extend='both')
for ax in axs[2:]:
    ax.contour(*centers,made,levels=[0],colors='k',linewidths=1,linestyles='--')
axs[2].text(0.92,0.92,'correction-enhanced',ha='right',va='top',transform=axs[2].transAxes)
axs[2].text(0.08,0.08,'correction-suppressed',ha='left',va='bottom',transform=axs[2].transAxes)
fig.colorbar(mcorr,loc='b',cols=(1,2),ticks=1,space=4,label='$\\Delta$Precipitation Rate (mm/hr)')
fig.format(abc=True,titleloc='l',grid=False)
pplt.show()
fig.save(os.path.join(FIGSDIR,'fig_6.jpg'))
print(round(fig.get_figwidth(),2))
```
