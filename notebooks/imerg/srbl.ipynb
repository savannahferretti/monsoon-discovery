```python
import os
import json
import warnings
import numpy as np
import pandas as pd
import xarray as xr
import proplot as pplt
from biascorrect import load_pairs,fit_qm
warnings.filterwarnings('ignore')
pplt.rc.update({
    'savefig.dpi':900,
    'savefig.bbox':'tight',
    'savefig.pad_inches':0.02,
    'tick.minor':False,
    'font.size':9,
    'label.size':9,
    'tick.labelsize':9,
    'title.size':9,
    'abc.size':9,
    'legend.fontsize':9,
    'suptitle.size':9,
    'leftlabelsize':9,
    'toplabelsize':9,
    'leftlabel.weight':'normal',
    'toplabel.weight':'normal',
    'reso':'xx-hi'})
```

```python
with open('../../scripts/configs.json','r',encoding='utf-8') as f:
    CONFIGS = json.load(f)
SPLITSDIR  = CONFIGS['filepaths']['splits']
MODELSDIR  = CONFIGS['filepaths']['models']
PREDSDIR   = CONFIGS['filepaths']['predictions']
EQUATIONS  = CONFIGS['experiments']['sr']['optimizedeqs']
SPLIT      = 'test'
NBINS      = 40
MINSAMPLES = 50
with open(os.path.join(SPLITSDIR,'stats.json'),'r',encoding='utf-8') as f:
    STATS = json.load(f)
REGISTRY   = {row['name']:json.loads(row['constants']) for _,row in pd.read_csv(os.path.join(MODELSDIR,'sr','optimized_equations.csv')).iterrows()}
NQUANTILES = 200
FITSPLITS  = ['train','valid']
FIGSDIR    = 'figs'
os.makedirs(FIGSDIR,exist_ok=True)
```

```python
def to_precip(exponent):
    return np.maximum(np.expm1(exponent),0.0)
```

```python
mapping = fit_qm(*load_pairs(SPLITSDIR,FITSPLITS),nquantiles=NQUANTILES)
c1,c2 = (REGISTRY['sr_bl_eq'][name] for name in ['c1','c2'])
lam   = STATS['tp_std']/STATS['bl_std']**3
bc    = STATS['bl_mean']-c1*STATS['bl_std']
beta  = STATS['tp_std']*c2

with xr.open_dataset(os.path.join(SPLITSDIR,f'{SPLIT}.h5'),engine='h5netcdf') as ds:
    bl,tp,pr = (ds[name].transpose('time','lat','lon').values.ravel() for name in ['bl','tp','pr'])
with xr.open_dataset(os.path.join(PREDSDIR,f'sr_bl_eq_{SPLIT}_predictions.nc')) as ds:
    savedtp = ds['tp'].squeeze().transpose('time','lat','lon').values.ravel()

finite = np.isfinite(bl)&np.isfinite(tp)
print(f'Max difference from saved SR-BL predictions: {np.nanmax(np.abs(to_precip(lam*(bl[finite]-bc)**3+beta)-savedtp[finite])):.2e} mm')
finite = np.isfinite(bl)&np.isfinite(pr)
bl,pr  = bl[finite],pr[finite]

blrange  = np.linspace(*np.percentile(bl,[1,99]),300)
predpr   = mapping(to_precip(lam*(blrange-bc)**3+beta))
edges    = np.linspace(*np.percentile(bl,[1,99]),NBINS+1)
centers  = 0.5*(edges[:-1]+edges[1:])
binidx   = np.clip(np.digitize(bl,edges)-1,0,NBINS-1)
counts   = np.bincount(binidx,minlength=NBINS)
binnedpr = np.where(counts>=MINSAMPLES,np.bincount(binidx,weights=pr,minlength=NBINS)/np.maximum(counts,1),np.nan)
```

```python
fig,ax = pplt.subplots(refwidth=2,refheight=2)
ax.format(grid=False,xlabel='$\\mathit{B_L}$ (m/s$^2$)',ylabel='Precipitation Rate (mm/hr)')
ax.plot(blrange,predpr,color=EQUATIONS['sr_bl_eq']['color'],linewidth=2,label='SR-BL',zorder=2)
ax.scatter(centers,binnedpr,color='gray',markersize=14,label='IMERG',zorder=3)
ax.axvline(bc,color='k',linestyle='--',linewidth=1,zorder=1)
ax.text(bc-0.01,0.8,'$\\mathit{B_c}$',ha='right',va='top',transform=ax.get_xaxis_transform())
ax.legend(loc='ul',ncols=1)
pplt.show()
fig.save(os.path.join(FIGSDIR,'fig_2.jpg'))
print(round(fig.get_figwidth(),2))
```
