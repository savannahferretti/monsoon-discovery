```python
import os
import json
import warnings
import numpy as np
import pandas as pd
import xarray as xr
import proplot as pplt
from matplotlib.ticker import StrMethodFormatter
from biascorrect import load_pairs,fit_qm
warnings.filterwarnings('ignore')
pplt.rc.update({
    'savefig.dpi':900,
    'savefig.bbox':'tight',
    'savefig.pad_inches':0.02,
    'tick.minor':False,
    'font.size':9,
    'label.size':9,
    'tick.labelsize':9,
    'title.size':9,
    'abc.size':9,
    'legend.fontsize':9,
    'suptitle.size':9,
    'leftlabelsize':9,
    'toplabelsize':9,
    'leftlabel.weight':'normal',
    'toplabel.weight':'normal',
    'reso':'xx-hi'})
```

```python
with open('../../scripts/configs.json','r',encoding='utf-8') as f:
    CONFIGS = json.load(f)
SPLITSDIR  = CONFIGS['filepaths']['splits']
WEIGHTSDIR = CONFIGS['filepaths']['weights']
MODELSDIR  = CONFIGS['filepaths']['models']
PREDSDIR   = CONFIGS['filepaths']['predictions']
FIELDVARS  = CONFIGS['experiments']['nn']['runs']['nn_gauss']['fieldvars']
SEEDS      = CONFIGS['experiments']['nn']['seeds']
LATRANGE   = CONFIGS['domain']['latrange']
LONRANGE   = CONFIGS['domain']['lonrange']
SPLIT      = 'test'
NBINS      = 35
MINSAMPLES = 50
with open(os.path.join(SPLITSDIR,'stats.json'),'r',encoding='utf-8') as f:
    STATS = json.load(f)
REGISTRY   = {row['name']:json.loads(row['constants']) for _,row in pd.read_csv(os.path.join(MODELSDIR,'sr','optimized_equations.csv')).iterrows()}
NQUANTILES = 200
FITSPLITS  = ['train','valid']
FIGSDIR    = 'figs'
os.makedirs(FIGSDIR,exist_ok=True)
```

```python
def to_precip(exponent):
    return np.maximum(np.expm1(exponent),0.0)
```

```python
mapping = fit_qm(*load_pairs(SPLITSDIR,FITSPLITS),nquantiles=NQUANTILES)
c3,c4,c5 = (REGISTRY['sr_atm_eq'][name] for name in ['c3','c4','c5'])
lam      = STATS['tp_std']*c3/STATS['thetae_std']**3
gamma    = c4*STATS['thetae_std']/STATS['thetaestar_std']
thetac   = STATS['thetae_mean']-gamma*STATS['thetaestar_mean']+c5*STATS['thetae_std']
kappa    = STATS['thetae_std']/STATS['rh_std']

with xr.open_dataset(os.path.join(SPLITSDIR,f'{SPLIT}.h5'),engine='h5netcdf') as ds:
    ntime,nlat,nlon = (ds.sizes[dim] for dim in ['time','lat','lon'])
    lat,lon,dsig    = ds['lat'].values,ds['lon'].values,ds['dsig'].values
    fields = np.stack([ds[name].transpose('time','lat','lon','sig').values.reshape(-1,dsig.size) for name in FIELDVARS],axis=1)
    tp     = ds['tp'].transpose('time','lat','lon').values.ravel()
    pr     = ds['pr'].transpose('time','lat','lon').values.ravel()
with xr.open_dataset(os.path.join(PREDSDIR,f'sr_atm_eq_{SPLIT}_predictions.nc')) as ds:
    savedtp = ds['tp'].squeeze().transpose('time','lat','lon').values.ravel()
kernels = []
for seed in SEEDS:
    with xr.open_dataset(os.path.join(WEIGHTSDIR,f'nn_gauss_{seed}_weights.nc'),engine='h5netcdf') as ds:
        kernels.append(ds['k'].values)

integrals   = dict(zip(FIELDVARS,(fields*np.mean(kernels,axis=0)[None,:,:]*dsig[None,None,:]).sum(axis=2).T))
moisture    = kappa*(integrals['rh']-STATS['rh_mean'])
instability = integrals['thetae']-gamma*integrals['thetaestar']-thetac
predtp      = to_precip(lam*np.maximum(moisture,instability)**3)
finite      = np.isfinite(moisture)&np.isfinite(instability)&np.isfinite(tp)&np.isfinite(pr)
print(f'Max difference from saved SR-ATM predictions: {np.nanmax(np.abs(predtp[finite]-savedtp[finite])):.2e} mm')

fracmap   = np.nanmean(np.where(finite,moisture>instability,np.nan).reshape(ntime,nlat,nlon),axis=0)
mlim,ilim = tuple(np.percentile(moisture[finite],[1,99])),tuple(np.percentile(instability[finite],[1,99]))
edges     = (np.linspace(*mlim,NBINS+1),np.linspace(*ilim,NBINS+1))
centers   = [0.5*(edge[:-1]+edge[1:]) for edge in edges]
binidx    = np.clip(np.digitize(moisture[finite],edges[0])-1,0,NBINS-1)*NBINS+np.clip(np.digitize(instability[finite],edges[1])-1,0,NBINS-1)
counts    = np.bincount(binidx,minlength=NBINS**2)
binnedpr  = np.where(counts>=MINSAMPLES,np.bincount(binidx,weights=pr[finite],minlength=NBINS**2)/np.maximum(counts,1),np.nan).reshape(NBINS,NBINS)
density   = np.log10(np.histogram2d(moisture[finite],instability[finite],bins=edges)[0].clip(1))
m,i       = np.linspace(*mlim,200),np.linspace(*ilim,200)
predgrid  = mapping(to_precip(lam*np.maximum(*np.meshgrid(m,i))**3))
diagonal  = [min(mlim[0],ilim[0]),max(mlim[1],ilim[1])]
```

```python
fig,axs = pplt.subplots([[1,2],[3,4]],refwidth=2.4,refheight=2.4,proj={4:'cyl'},share=False,wspace=1.2,hspace=8)
axs[2].set_box_aspect((LATRANGE[1]-LATRANGE[0])/(LONRANGE[1]-LONRANGE[0]))
m0 = axs[0].pcolormesh(m,i,predgrid,cmap='ColdHot_r',cmap_kw={'left':0.52},vmin=0,vmax=4,levels=18,extend='max')
m1 = axs[1].pcolormesh(centers[0],centers[1],binnedpr.T,cmap='ColdHot_r',cmap_kw={'left':0.52},vmin=0,vmax=4,levels=18,extend='max')
m2 = axs[2].pcolormesh(centers[0],centers[1],density.T,cmap='Grays',vmin=0,vmax=5)
m3 = axs[3].pcolormesh(lon,lat,fracmap,cmap='ColdHot_r',vmin=0,vmax=1)
for ax in axs[:3]:
    ax.plot(diagonal,diagonal,'k--',linewidth=1)
axs[0].text(0.08,0.92,'instability-dominated',ha='left',va='top',transform=axs[0].transAxes)
axs[0].text(0.92,0.08,'moisture-dominated',ha='right',va='bottom',transform=axs[0].transAxes)
axs[:3].format(xlabel='$\\mathit{M}$ (K)',xlabelpad=2,xlim=mlim,ylim=ilim)
axs[0].format(title='SR-ATM',ylabel='$\\mathit{I}$ (K)')
axs[1].format(title='IMERG',yticklabels=[])
axs[2].format(title='Sample Density',ylabel='$\\mathit{I}$ (K)')
axs[3].format(title='Pathway Dominance by Location',coast=True,lonlim=LONRANGE,lonlines=[65,75,85],lonlabels='b',latlim=LATRANGE,latlines=[10,15,20],latlabels='r',grid=False)
axs[2].colorbar(m2,loc='b',label='log$_{10}$($\\mathit{N}$)')
axs[3].colorbar(m3,loc='b',ticks=0.2,label='Fraction of Timesteps with $\\mathit{M} > \\mathit{I}$')
fig.format(abc=True,titleloc='l')

fig.canvas.draw()
top    = [ax.get_position() for ax in axs[:2]]
bottom = [ax.get_position() for ax in axs[2:]]
middle = 0.5*(min(pos.y0 for pos in top)+max(pos.y1 for pos in bottom))
cax    = fig.add_axes([top[0].x0,middle+0.06/fig.get_figheight(),top[1].x1-top[0].x0,0.18/fig.get_figheight()])
cb     = fig.colorbar(m0,cax=cax,orientation='horizontal',label='Precipitation Rate (mm/hr)',extend='max')
cb.ax.xaxis.set_major_formatter(StrMethodFormatter('{x:g}'))
pplt.show()
fig.save(os.path.join(FIGSDIR,'fig_4.jpg'))
print(round(fig.get_figwidth(),2))
```
